```python
# 2. Membuat tabel frekuensi berdasarkan kolom Grade
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")

print("--Tabel Frekuensi Grade--")
print(datafrq)
```

```python
# a. Grafik Garis (Line Chart)
datafrq.plot(kind='line', marker='o', title='Grafik Garis Frekuensi Grade', color='yellow')
plt.ylabel('Frekuensi')
plt.show()

print("\n")

print("\n")

```

```python
# b. Grafik Batang (Bar Chart)
datafrq.plot(kind='bar', color='pink', title='Grafik Batang Frekuensi Grade')
plt.ylabel('Frekuensi')
plt.xticks(rotation=0)
plt.show()
```

```python
# c. Grafik Lingkaran (Pie Chart)
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False,
             title='Persentase Distribusi Nilai (Grade)')
plt.ylabel('') # Menghilangkan label Y di samping lingkaran
plt.show()
```

```python
# 4. Menghitung Descriptive Statistics

dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['Sample Variance'] = dt.var()
stats['Range'] = dt.max() - dt.min()
stats['Skewness'] = dt.skew()
stats['Kurtosis'] = dt.kurtosis()

print("--Statistik Deskriptif Final Score--")
print(stats)
```
